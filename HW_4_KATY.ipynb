```python
import random
```

```python
def generate_password(lenght=10):

  str1 = '123456789'
  str2 = 'abcdefghijklmnopqrstuvwxyz'
  str3 = str2.upper()
  str4 = '!@#$%^&*'
  list = str1 + str2 + str3 + str4
  password = ''.join((random.choice(list)) for _ in range(lenght))
  return password

print(generate_password(10))

```

```python
import re
from collections import Counter
```

```python
text = "Моя семья живет в пятикомнатной квартире, которая находится в центре города. У каждого члена семьи есть своя спальня. Гостиная – самая большая комната в нашей квартире, а моя спальня имеет самую маленькую площадь. Несмотря на то, что в моей комнате очень мало места, здесь есть все необходимое: письменный стол, стул, шкаф для одежды и кровать. Мне нравится то, что кровать стоит рядом со стеной, потому что в стене установлена розетка для зарядки телефона."
```

```python
def analyze_text(text: str):
  words = re.findall(r'\b\w+\b', text.lower())


if not words:
  return {
            "total_words": 0,
            "unique_words": 0,
            "most_common": None,
            "avg_length": 0
        }

total_words = len(words)
unique_words = len(set(words))

    # Находим самое частое слово
counter = Counter(words)
max_count = max(counter.values())
most_common_words = [word for word, count in counter.items() if count == max_count]

    # Средняя длина слова
avg_length = sum(len(word) for word in words) / total_words
return {
        "total_words": total_words,
        "unique_words": unique_words,
        "most_common": most_common_words,
        "avg_length": avg_length
    }

if __name__ == "__main__":
    try:
        text = input("Введите текст: ").strip()
        result = analyze_text(text)

        print(f"Количество слов: {result['total_words']}")
        print(f"Количество уникальных слов: {result['unique_words']}")
        if result['most_common']:
            print(f"Самое частое слово(а): {', '.join(result['most_common'])}")
        else:
            print("Самое частое слово: отсутствует")
        print(f"Средняя длина слова: {result['avg_length']:.2f}")
    except Exception as e:
        print(f"Ошибка: {e}")
```
